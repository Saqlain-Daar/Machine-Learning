```python
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split

# Assuming data_cleaned is your DataFrame
X = data_cleaned.drop('price', axis=1)  # Features
y = data_cleaned['price']  # Target

# Step 1: Handle categorical variables with one-hot encoding
X_encoded = pd.get_dummies(X, drop_first=True)  # drop_first to avoid the dummy variable trap

# Step 2: Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X_encoded, y, test_size=0.2, random_state=42)

# Step 3: Initialize the StandardScaler
scaler = StandardScaler()

# Step 4: Scale the features (X)
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Verify scaling (optional)
print(X_train_scaled[:5])  # Print the first 5 rows of scaled data
```

```python
print(data_cleaned)
```

```python
from sklearn.linear_model import LinearRegression

# Initialize the model
model = LinearRegression()

# Train the model on the scaled training data
model.fit(X_train_scaled, y_train)

# Check the R^2 score (good indicator of model fit)
print(f"Training R^2 score: {model.score(X_train_scaled, y_train)}")
print(f"Testing R^2 score: {model.score(X_test_scaled, y_test)}")
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

# Predicting on the test data
y_pred = model.predict(X_test_scaled)

# Calculate the Mean Absolute Error (MAE)
mae = mean_absolute_error(y_test, y_pred)
print(f"Mean Absolute Error: {mae}")

# Calculate the Mean Squared Error (MSE)
mse = mean_squared_error(y_test, y_pred)
print(f"Mean Squared Error: {mse}")

# Calculate the Root Mean Squared Error (RMSE)
rmse = np.sqrt(mse)
print(f"Root Mean Squared Error: {rmse}")
```

```python
from sklearn.ensemble import RandomForestRegressor

# Initialize the model
model = RandomForestRegressor(n_estimators=100, random_state=42)

# Train the model on the scaled training data
model.fit(X_train_scaled, y_train)

# Check the R^2 score (good indicator of model fit)
print(f"Training R^2 score: {model.score(X_train_scaled, y_train)}")
print(f"Testing R^2 score: {model.score(X_test_scaled, y_test)}")
```

```python
import xgboost as xgb
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

# Initialize the XGBoost model
model_xgb = xgb.XGBRegressor(random_state=42)

# Train the model on the scaled training data
model_xgb.fit(X_train_scaled, y_train)

# Check the R^2 score on training and testing data
train_score = model_xgb.score(X_train_scaled, y_train)
test_score = model_xgb.score(X_test_scaled, y_test)

print(f"Training R^2 score: {train_score}")
print(f"Testing R^2 score: {test_score}")

# Predict on the test data
y_pred_xgb = model_xgb.predict(X_test_scaled)

# Calculate the Mean Absolute Error (MAE)
mae_xgb = mean_absolute_error(y_test, y_pred_xgb)
print(f"Mean Absolute Error: {mae_xgb}")

# Calculate the Mean Squared Error (MSE)
mse_xgb = mean_squared_error(y_test, y_pred_xgb)
print(f"Mean Squared Error: {mse_xgb}")

# Calculate the Root Mean Squared Error (RMSE)
rmse_xgb = np.sqrt(mse_xgb)
print(f"Root Mean Squared Error: {rmse_xgb}")
```
